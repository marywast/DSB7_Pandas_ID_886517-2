```python
import pandas as pd
```

```python
data = "../data/auto.json"
```

## **CarNumber в качестве индекса**

```python
df = pd.read_json(data, orient="records")
```

```python
df = df.set_index("CarNumber")
```

## **Выборки**

### **Строки, где штрафы больше 2 100**

```python
df[df["Fines"] > 2100]
```

### **Cтроки, где штрафы больше 2 100, а возврат (Refund) равен 2**

```python
df[(df["Fines"] > 2100) & (df["Refund"] == 2)]
```

### **Строки, где модель входит в список: ['Focus', 'Corolla']**

```python
df[df["Model"].isin(["Focus", "Corolla"])]
```

### **Строки, где номер автомобиля входит в список: ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']**

```python
df[
    df.index.isin(
        ["Y7689C197RUS", "92928M178RUS", "7788KT197RUS", "H115YO163RUS", "X758HY197RUS"]
    )
]
```

## **Агрегации по Make и Model**

### **Медианные значения штрафов, сгруппированные по Make**

```python
df.groupby("Make")["Fines"].median()
```

```python
df.groupby(['Make', 'Model']).agg('Fines').count()
```

### **Медианные значения штрафов, сгруппированные по Make и Model**

```python
df.groupby(["Make", "Model"])["Fines"].median()
```

### **Количество штрафов, сгруппированное по Make и Model**

```python
df.groupby(["Make", "Model"])["Fines"].agg(["count"])
```

### **Минимальные и максимальные штрафы, сгруппированные по Make и Model**

```python
df.groupby(["Make", "Model"])["Fines"].agg(["min", "max"])
```

### **Стандартное отклонение штрафов, сгруппированное по Make и Model**

```python
df.groupby(["Make", "Model"])["Fines"].agg(["std"])
```

## **Агрегации по номерам автомобилей**

### **Номера автомобилей, сгруппированные по количеству штрафов в порядке убывания**

```python
car_counts = df.groupby(df.index).size().sort_values(ascending=False)
```

```python
car_counts
```

### **Все строки, соответствующие топ-1 номеру**


```python
df[df.index == car_counts.idxmax()]
```

### **Номера автомобилей, сгруппированные по сумме штрафов в порядке убывания**

```python
fines_sum = df.groupby(df.index)["Fines"].sum().sort_values(ascending=False)
```

```python
fines_sum
```

### **Все строки, соответствующие топ-1 номеру по сумме штрафов**

```python
df[df.index == fines_sum.idxmax()]
```

### **«Есть ли номера автомобилей, привязанные к разным моделям?»**

```python
df.groupby(df.index)["Model"].nunique()[lambda x: x > 1]
```
