```python
import pandas as pd
```

```python
data = "../data/feed-views.log"
```

## **Датафрейм views**

```python
views = pd.read_csv(data, names=["datetime", "user"], sep="\t")

views["datetime"] = pd.to_datetime(views["datetime"])
views["year"] = views["datetime"].dt.year
views["month"] = views["datetime"].dt.month
views["day"] = views["datetime"].dt.day
views["hour"] = views["datetime"].dt.hour
views["minute"] = views["datetime"].dt.minute
views["second"] = views["datetime"].dt.second
```

## **Столбец daytime**

```python
bins = [0, 4, 7, 11, 17, 20, 24]
labels = ["night", "early morning", "morning", "afternoon", "early evening", "evening"]
views["daytime"] = pd.cut(views["hour"], bins=bins, labels=labels, right=False)

views = views.set_index("user")
```

## **Подсчет элементов в датафрейме**

```python
views.count()
```

```python
views.value_counts("daytime")
```

## **Сортировка значений в датафрейме одновременно по часам, минутам и секундам по возрастанию**

```python
views.sort_values(["hour", "minute", "second"])
```

## **Минимум и максимум для часов и мода для категорий daytime**

## **Максимальный час для строк, где время суток - night**

```python
max = views[views["daytime"] == "night"]["hour"].max()
```

## **Минимальный час для строк, где время суток - morning**

```python
min = views[views["daytime"] == "morning"]["hour"].min()
```

## **Пользователи, которые посещали страницу в найденные часы**



```python
views[views["hour"] == max].head(1)
```

```python
views[views["hour"] == min].head(1)
```

```python
views["hour"].mode()
```

```python
views["daytime"].mode()
```
