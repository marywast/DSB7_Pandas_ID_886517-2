```python
import pandas as pd
```

```python
data = "../data/feed-views.log"
```

## **read_csv**

```python
df = pd.read_csv(
    data,
    skiprows=[2, 3],
    skipfooter=2,
    engine="python",
    names=["datetime", "user"],
    sep="\t",
    index_col="datetime",
)
```

## **datetime в date_time**

```python
df.index.name = "date_time"
```

## **to_csv**

```python
df.to_csv("feed-views-semicolon.log", sep=";")
```
