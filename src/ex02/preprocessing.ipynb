```python
import pandas as pd
```

```python
data = "../data/auto.csv"
```

## **Чтение файла и ID в качестве индексного столбца**

```python
df = pd.read_csv(data)
df = df.set_index("ID")
```

## **count()**

```python
df.count()
```

## **Удаление дубликатов в столбцах CarNumber, Make_n_Model и Fines**

```python
df.duplicated(subset=["CarNumber", "Make_n_model", "Fines"])
```

```python
df = df.drop_duplicates(subset=["CarNumber", "Make_n_model", "Fines"], keep="last")
```

```python
df.count()
```

## **Пропуски**

```python
df.isna().sum()
```

### **Удаление History**

```python
df = df.dropna(axis=1, thresh=500)
```

```python
df.isna().sum()
```

### **Замена пропусков в столбце Refund предыдущим значением**

```python
df["Refund"] = df["Refund"].fillna(method="ffill")
```

```python
df.isna().sum()
```

### **Замена пропусков в столбце Fines средним значением этого столбца (не учитывая NA/NULL при вычислении среднего)**

```python
df["Fines"] = df["Fines"].fillna(df["Fines"].mean())
```

```python
df.isna().sum()
```

### **Разделение столбца Make_n_Model на столбцы Make и Model**

```python
df["Make"] = df["Make_n_model"].apply(lambda x: x.split()[0])
df["Model"] = df["Make_n_model"].apply(
    lambda x: x.split()[1] if len(x.split()) > 1 else None
)
```

```python
df = df.drop(columns=["Make_n_model"])
```

```python
df.to_json("auto.json", orient="records", indent=2)
```
